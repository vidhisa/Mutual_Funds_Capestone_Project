```python
import os
import requests
import pandas as pd

print("1. Setting up the web link and the folder path...")
url = "https://api.mfapi.in/mf/125497"

# Where we want to save our finished file
folder = "data/raw"
file_path = "data/raw/live_hdfc_nav_history.csv"

print("2. Requesting the data from the website...")
# This goes out to the internet and grabs the raw text/JSON data from the API
response = requests.get(url)

# Turn that raw internet data into something Python can understand (a regular dictionary)
all_data = response.json()

print("3. Pulling out the historical data records...")
# The website sends back two things: "meta" (general info) and "data" (the actual rows of NAV history)
# We only want the rows of dates and NAV values, which are stored inside "data"
raw_rows = all_data["data"]

print("4. Turning the data rows into a clean spreadsheet table...")
# Convert that raw list of rows into a beautiful Pandas DataFrame table
df = pd.DataFrame(raw_rows)

print("5. Making sure our folder exists on the computer...")
# If the "data/raw" folder isn't there yet, this command creates it automatically
os.makedirs(folder, exist_ok=True)

print("6. Saving the table as a clean CSV file...")
# Save our new table directly into that folder as a CSV spreadsheet
# index=False prevents pandas from adding an extra column of row numbers (0, 1, 2...)
df.to_csv(file_path, index=False)

print("🎉 Success! The file has been saved perfectly inside data/raw/")
print("-" * 50)

# Show a quick preview of the top rows on the screen
print("👀 Here is a quick look at the first 5 rows of your file:")
print(df.head())
```

```python
import os
import requests
import pandas as pd

# 1. Create a simple list of the 5 schemes with their codes and names
schemes = {
    "119551": "sbi_bluechip",
    "120503": "icici_bluechip",
    "118632": "nippon_large_cap",
    "119092": "axis_bluechip",
    "120841": "kotak_bluechip"
}

# Make sure your data/raw folder exists before we start downloading
os.makedirs("data/raw", exist_ok=True)

print("🚀 Starting the automated download for all 5 mutual funds...\n")

# 2. Tell Python to loop through each scheme code one by one
for code, name in schemes.items():
    print(f"🔄 Processing: {name.upper()} (Code: {code})...")
    
    # Create the customized website link for this specific scheme code
    web_link = f"https://api.mfapi.in/mf/{code}"
    
    # Download the data from the internet
    web_response = requests.get(web_link)
    raw_json_data = web_response.json()
    
    # Extract just the rows of historical dates and NAV values
    just_the_rows = raw_json_data["data"]
    
    # Turn the rows into a beautiful table spreadsheet
    nav_table = pd.DataFrame(just_the_rows)
    
    # Create a clean file path to save this specific fund
    # This will create files like: data/raw/live_sbi_bluechip_nav.csv
    file_path = f"data/raw/live_{name}_nav.csv"
    
    # Save the table as a clean CSV file
    nav_table.to_csv(file_path, index=False)
    
    # Print a quick success status message for this file
    print(f"   💾 Saved! Rows: {nav_table.shape[0]} -> Path: {file_path}\n")

print("==================================================")
print("🎉 All 5 mutual fund datasets downloaded and saved perfectly!")
print("==================================================")
```
